```python
# Count the number of common elements in 2 sorted lists

a = [1,2,3]
b = [3,3,4,5]

def common_elements(a, b):
    out = 0
    i =0
    j =0

    while i < len(a) and j < len(b):
        if a[i] < b[j]:
            i += 1
        
        elif a[i] > b[j]:
            j += 1

        else:
            out += 1
            i += 1
            j += 1

    return out

common_elements(a,b)
```
