```python
# 49. Group Anagrams
"""
Given an array of strings strs, group the together. You can return the answer in any order.

Example 1:
Input: strs = ["eat","tea","tan","ate","nat","bat"]
Output: [["bat"],["nat","tan"],["ate","eat","tea"]]
Explanation:
    There is no string in strs that can be rearranged to form "bat".
    The strings "nat" and "tan" are anagrams as they can be rearranged to form each other.
    The strings "ate", "eat", and "tea" are anagrams as they can be rearranged to form each other.

Example 2:
Input: strs = [""]
Output: [[""]]

Example 3:
Input: strs = ["a"]
Output: [["a"]]

Constraints:
    1 <= strs.length <= 104
    0 <= strs[i].length <= 100
    strs[i] consists of lowercase English letters.
"""
```

```python
from rich import print as rprint

strs = ["eat","tea","tan","ate","nat","bat"]
expected = [["bat"],["nat","tan"],["ate","eat","tea"]]
```

```python
# O(n*klog(k))

word_map = {} # sorted_str -> [strs]
for word in strs:
    sorted_chars = sorted(word)
    sorted_word = "".join(sorted_chars)
    if sorted_word in word_map:
        word_map[sorted_word].append(word)
    else:
        word_map[sorted_word] = [word]

out = [str_list for str_list in word_map.values()] # return


assert set(frozenset(group) for group in out) == set(frozenset(group) for group in expected), f"expected {expected} and output {out} do not match"
rprint(out)

```

```python
# cleaner version with defaultdict - still O(n*klog(k)))
from collections import defaultdict

word_map = defaultdict(list)
for word in strs:
    sorted_chars = sorted(word)
    sorted_word = "".join(sorted_chars)
    word_map[sorted_word].append(word)

out = [str_list for str_list in word_map.values()] # Return

assert set(frozenset(group) for group in out) == set(frozenset(group) for group in expected), f"expected {expected} and output {out} do not match"
rprint(out)
```

```python
# O(n*k)
from collections import defaultdict

res = defaultdict(list)
for word in strs:
    count = [0] * 26
    for char in word:
        count[ord(char) - ord("a")] += 1

    print(f"count for {word}:", count)

    res[tuple(count)].append(word)   

out =[str_list for str_list in res.values()] # return

assert set(frozenset(group) for group in out) == set(frozenset(group) for group in expected), f"expected {expected} and output {out} do not match"
rprint(out)
```
