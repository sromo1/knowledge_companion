```python
## 347. Top K Frequent Elements
"""
Given an integer array nums and an integer k, return the k most frequent elements. You may return the answer in any order.

Example 1:
Input: nums = [1,1,1,2,2,3], k = 2
Output: [1,2]

Example 2:
Input: nums = [1], k = 1
Output: [1]

Example 3:
Input: nums = [1,2,1,2,1,2,3,1,3,2], k = 2
Output: [1,2]

Constraints:
    1 <= nums.length <= 105
    -104 <= nums[i] <= 104
    k is in the range [1, the number of unique elements in the array].
    It is guaranteed that the answer is unique.

Follow up: Your algorithm's time complexity must be better than O(n log n), where n is the array's size.
"""
```

```python
from rich import print as rprint

nums = [1,2,1,2,1,2,3,1,3,2]
k = 2
expected = [1,2]
```

```python
# Brute force:
from collections import Counter

num_counts = Counter(nums)
num_counts_sorted = list(sorted(num_counts.items(), key=lambda item: item[0]))
print('sorted num counts:', num_counts_sorted)
out = [entry[0] for entry in num_counts_sorted[:k]]

assert out == expected
rprint('output:', out)
```

```python
# Practically efficient 
from collections import Counter
num_counts = Counter(nums)
top_k_nums = num_counts.most_common(k)
out = [entry[0] for entry in top_k_nums] # return

assert out == expected
rprint('output:', out)
```

```python
# Theoretically optimal
from collections import Counter
num_counts = Counter(nums)

n = len(nums)
buckets = [[] for _ in range(n+1)]
print('empty buckets:', buckets)

for num, freq in num_counts.items():
    buckets[freq].append(num)
print('filled buckets:', buckets)
out = []
for i in range(n, 0, -1):
    for num in buckets[i]:
        out.append(num)
        if len(out)==k:
            assert out==expected
            rprint('output:', out) # return
```

```python
# heaps 
from collections import Counter
import heapq

num_counts = Counter(nums)

heap = []
for num, freq in num_counts.items():
    heapq.heappush(heap, (freq, num)) # Push to heap, lowest freq remains first
    if len(heap) > k:
        heapq.heappop(heap) # Pop lowest freq (first item)

print('filled heap:', heap)
out = [item[1] for item in heap]

assert out == expected
rprint('output:', out)
```
