```python
# 387. First Unique Character in a String
"""
Given a string s, find the first non-repeating character in it and return its index. If it does not exist, return -1.

Example 1:
Input: s = "leetcode"
Output: 0
Explanation:
The character 'l' at index 0 is the first character that does not occur at any other index.

Example 2:
Input: s = "loveleetcode"
Output: 2

Example 3:
Input: s = "aabb"
Output: -1

Constraints:
    1 <= s.length <= 105
    s consists of only lowercase English letters.
"""

from rich import print as rprint

s = "loveleetcode"
expected = 2
```

```python
# Time: O(n)
# Space: O(1)
# Fastest as Counter is implemented in C
from collections import Counter

counts = Counter(s)
print("counts:", counts)

out = -1
for i, char in enumerate(s):
    if counts[char] == 1:
        out = i # return
        break

assert out == expected, f'expected {expected} but got {out}'
rprint(out)
```

```python
# An equivalent solution using defaultdict
from collections import defaultdict

counts = defaultdict(int)
print(counts)

for char in s:
    counts[char] += 1
print("counts:", counts)

out = -1
for i, char in enumerate(s):
    if counts[char] == 1:
        out = i # return
        break

assert out == expected, f"Expected {expected}, but got {out}"
rprint(out)
```

```python
lookup = {} # char -> [counts, first_index]
for i, char in enumerate(s):
    if char in lookup:
        lookup[char][0] += 1
    else:
        lookup[char] = [1, i]
print("lookup:", lookup)

out = -1
for counts, first_index in lookup.values():
    if counts == 1:
        out = first_index # return
        break

assert out == expected, f"Expected {expected}, but got {out}"
rprint(out)
```
