```python
# 1.1 Is Unique: Implement an algorithm to determine if a string has all unique characters. What if you cannot use additional data structures?

def isUnique(string:str) -> bool:
    ''' O(n) '''
    char_set = set()
    for c in string:
        if c in char_set:
            return False
        char_set.add(c)
    return True
    

def isUnique_noDS(string:str) -> bool:
    ''' O(n^2) '''
    if len(string) <= 1: # Handle edge case
        return True
    
    for i in range(0, len(string)-1):
        for j in range(i+1, len(string)):
            if string[i] == string[j]:
                return False
    return True

def isUnique_noDS_optimized(string:str) -> bool:
    ''' O(n) '''
    checker = 0
    for c in string:
        val = ord(c) - ord('a')
        if (checker & (1<<val)) > 0:
            return False
        checker |= (1<<val)
    return True

test_cases = {
    'abc'   : True,
    'abca'  : False,
    ''      : True,
    's'     : True,
    'abcc'  : False,
    'aab'   : False,
}

print('test isUnique:')
test_solution(isUnique, test_cases)

print('\ntest isUnique_noDS:')
test_solution(isUnique_noDS, test_cases)

print('\ntest isUnique_noDS_optimized:')
test_solution(isUnique_noDS_optimized, test_cases)
```
