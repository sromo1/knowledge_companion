```python
def test_solution(fun:callable, test_cases: dict):
    for i, test_case in enumerate(test_cases.items()):
        input = test_case[0]

        if type(input) == tuple:
            evaluation = fun(*input)
            
        else: # Base case - single input
            evaluation = fun(input)
        output = test_case[1]
        
        
        assert evaluation == output, f"failed case {i}: {input} expected {output}, received {evaluation}"
    print(f'All {len(test_cases.keys())} tests passed')
```

```python
# 1.1 Is Unique: Implement an algorithm to determine if a string has all unique characters. What if you cannot use additional data structures?

def isUnique(string:str) -> bool:
    ''' O(n) '''
    char_set = set()
    for c in string:
        if c in char_set:
            return False
        char_set.add(c)
    return True
    

def isUnique_noDS(string:str) -> bool:
    ''' O(n^2) '''
    if len(string) <= 1: # Handle edge case
        return True
    
    for i in range(0, len(string)-1):
        for j in range(i+1, len(string)):
            if string[i] == string[j]:
                return False
    return True

def isUnique_noDS_optimized(string:str) -> bool:
    ''' O(n) '''
    checker = 0
    for c in string:
        val = ord(c) - ord('a')
        if (checker & (1<<val)) > 0:
            return False
        checker |= (1<<val)
    return True

test_cases = {
    'abc'   : True,
    'abca'  : False,
    ''      : True,
    's'     : True,
    'abcc'  : False,
    'aab'   : False,
}

print('test isUnique:')
test_solution(isUnique, test_cases)

print('\ntest isUnique_noDS:')
test_solution(isUnique_noDS, test_cases)

print('\ntest isUnique_noDS_optimized:')
test_solution(isUnique_noDS_optimized, test_cases)
```

```python
 # Given a string, determine if it is a permutation of a palindrome.

def bitSum(n: int) -> 0:
    counter = 0
    while (n > 0):
        n &= (n-1)
        counter += 1
    return counter

def is_palindrome_permutation(string:str) -> bool:
    checker = 0
    for c in string:
        val = ord(c) - ord('a')
        checker ^= (1<<val)
    if bitSum(checker) <= 1:
        return True
    else:
        return False
    
    
test_cases = {
    'abc'   : False,
    'aba'   : True,
    ''      : True, 
    's'     : True,
    'acc'   : True,
    'aab'   : True,
    'aa'    : True,
    'aabc'  : False,
}

print('test is_palindrome_permutation:')
test_solution(is_palindrome_permutation, test_cases)
```

```python
# 1.2 Check Permutation: Given two strings, write a method to decide if one is a permutation of the other.
test_cases = {
    ('abc', 'bca')   : True,
    ('a', 'a')       : True,
    ('', '')         : True,
    ('a', 'ab')      : False,
    ('ab', 'a')      : False,
    ('ad', 'bc')     : False,
}

def checkPermutation(str_1, str_2) -> bool:
    """" O(n) """
    from collections import Counter
    if len(str_1) != len(str_2):
        return False
    return Counter(str_1) == Counter(str_2)

print('test checkPermutation:')
test_solution(checkPermutation, test_cases)


def checkPermutation_noDS(str_1, str_2) -> bool:
    """O(nlog(n))"""
    l1 = list(str_1)
    l2 = list(str_2)

    l1.sort()
    l2.sort()

    return l1 == l2

print('test checkPermutation_noDS:')
test_solution(checkPermutation_noDS, test_cases)
```
